```
import pandas as pd
import joblib

# Helper Function (Can't be accessed from the other notebook)
def basic_cleaning(x_input):
    x_out = x_input.copy()
    x_out['was_contacted'] = (x_out['pdays'] != -1).astype(int)
    x_out['pdays'] = x_out['pdays'].replace(-1, 0)
    return x_out

# Load model
model = joblib.load('../models/modelo_final.joblib')

# Load competition data
df_competition = pd.read_pickle('../data/bank_competition_14.pkl')

# Predict
predictions = model.predict(df_competition)
y_pred_labels = pd.Series(predictions).map({1: 'yes', 0: 'no'})

# Save
y_pred_labels.to_csv('../outputs/predicciones.csv', index=False, header=['deposit'])
print("Done.")
print(y_pred_labels.value_counts())
```
